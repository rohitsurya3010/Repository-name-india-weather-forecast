```python
rainfall_path = r"C:\Users\rohit\Downloads\archive (4)\daily-rainfall-at-state-level.csv"

rainfall = pd.read_csv(rainfall_path)

print("Shape:", rainfall.shape)
print("Columns:", rainfall.columns.tolist())

rainfall.head()
```

```python
rainfall["date"] = pd.to_datetime(rainfall["date"])

print("Date range:")
print(rainfall["date"].min(), "to", rainfall["date"].max())

print("\nNumber of states:")
print(rainfall["state_name"].nunique())

print("\nStates:")
print(rainfall["state_name"].unique())

print("\nMissing values:")
print(rainfall.isnull().sum())
```

```python
rainfall_ml = rainfall.dropna(subset=["actual", "rfs"]).copy()

rainfall_ml["forecast_error"] = (
    rainfall_ml["actual"] - rainfall_ml["rfs"]
).abs()

# Forecast bust = error >= 20 mm
rainfall_ml["forecast_bust"] = (
    rainfall_ml["forecast_error"] >= 20
).astype(int)

print("ML dataset shape:", rainfall_ml.shape)

print("\nBust distribution:")
print(rainfall_ml["forecast_bust"].value_counts())

print("\nBust percentage:")
print(rainfall_ml["forecast_bust"].mean() * 100)
```

```python
# Approximate geographic center of each Indian state/UT
state_coords = {
    "Andhra Pradesh": (15.9, 79.7),
    "Arunachal Pradesh": (28.2, 94.7),
    "Assam": (26.2, 92.9),
    "Bihar": (25.9, 85.3),
    "Chhattisgarh": (21.3, 82.0),
    "Goa": (15.3, 74.1),
    "Gujarat": (22.3, 71.2),
    "Haryana": (29.1, 76.1),
    "Himachal Pradesh": (31.1, 77.2),
    "Jharkhand": (23.6, 85.3),
    "Karnataka": (15.3, 75.7),
    "Kerala": (10.5, 76.3),
    "Madhya Pradesh": (23.5, 78.0),
    "Maharashtra": (19.7, 75.7),
    "Manipur": (24.7, 93.9),
    "Meghalaya": (25.5, 91.4),
    "Mizoram": (23.2, 92.9),
    "Nagaland": (26.2, 94.6),
    "Odisha": (20.5, 84.4),
    "Punjab": (31.1, 75.3),
    "Rajasthan": (27.0, 74.2),
    "Sikkim": (27.5, 88.5),
    "Tamil Nadu": (11.1, 78.7),
    "Telangana": (18.1, 79.0),
    "Tripura": (23.9, 91.9),
    "Uttar Pradesh": (26.8, 80.9),
    "Uttarakhand": (30.1, 79.2),
    "West Bengal": (23.0, 87.9),
    "Delhi": (28.6, 77.2),
    "Jammu and Kashmir": (33.8, 76.6),
    "Ladakh": (34.2, 77.6),
    "Puducherry": (11.9, 79.8),
    "Chandigarh": (30.7, 76.8),
    "Dadra and Nagar Haveli and Daman and Diu": (20.2, 73.0),
    "Andaman and Nicobar Islands": (11.7, 92.7),
    "Lakshadweep": (10.6, 72.6)
}

print("States in coordinate dictionary:", len(state_coords))
print("States in rainfall data:", rainfall_ml["state_name"].nunique())
```

```python
from sklearn.neighbors import NearestNeighbors

# GFS grid points
grid = nwp_data[
    ["latitude", "longitude"]
].drop_duplicates().reset_index(drop=True)

# State coordinates
state_df = pd.DataFrame(
    [
        {
            "state_name": state,
            "state_lat": lat,
            "state_lon": lon
        }
        for state, (lat, lon) in state_coords.items()
    ]
)

# IMPORTANT: use numpy values so column names don't matter
nn = NearestNeighbors(n_neighbors=1)

nn.fit(
    grid[["latitude", "longitude"]].values
)

distances, indices = nn.kneighbors(
    state_df[["state_lat", "state_lon"]].values
)

# Attach nearest GFS grid
state_df["grid_latitude"] = (
    grid.iloc[indices.flatten()]["latitude"].values
)

state_df["grid_longitude"] = (
    grid.iloc[indices.flatten()]["longitude"].values
)

state_df["grid_distance"] = distances.flatten()

print("Mapped states:", len(state_df))

state_df.head(10)
```

```python

# Work with valid historical forecast-observation pairs
hist = rainfall.dropna(subset=["actual", "rfs"]).copy()

# Date features
hist["date"] = pd.to_datetime(hist["date"])
hist["month"] = hist["date"].dt.month
hist["year"] = hist["date"].dt.year

# Absolute forecast error
hist["forecast_error"] = (
    hist["actual"] - hist["rfs"]
).abs()

# Forecast bust: error >= 20 mm
hist["forecast_bust"] = (
    hist["forecast_error"] >= 20
).astype(int)

# Sort chronologically
hist = hist.sort_values(
    ["state_name", "date"]
).reset_index(drop=True)

# Historical behavior WITHOUT using the current day's error
hist["previous_error_mean"] = (
    hist.groupby("state_name")["forecast_error"]
    .transform(lambda x: x.shift(1).rolling(30, min_periods=5).mean())
)

hist["previous_bust_rate"] = (
    hist.groupby("state_name")["forecast_bust"]
    .transform(lambda x: x.shift(1).rolling(30, min_periods=5).mean())
)

# Fill unavailable early-history values
hist["previous_error_mean"] = hist["previous_error_mean"].fillna(
    hist["forecast_error"].median()
)

hist["previous_bust_rate"] = hist["previous_bust_rate"].fillna(
    hist["forecast_bust"].mean()
)

print("Historical ML dataset:", hist.shape)

print("\nFeatures:")
print([
    "rfs",
    "normal",
    "deviation",
    "month",
    "previous_error_mean",
    "previous_bust_rate"
])

print("\nBust rate:")
print(hist["forecast_bust"].mean() * 100)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Create a clean ML dataset
ml = hist.copy()

# Features available before knowing the actual rainfall
feature_columns = [
    "state_name",
    "rfs",
    "normal",
    "deviation",
    "month",
    "previous_error_mean",
    "previous_bust_rate"
]

X = ml[feature_columns].copy()
y = ml["forecast_bust"].copy()

# Remove any remaining missing values
valid = X.notna().all(axis=1) & y.notna()

X = X[valid]
y = y[valid]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget percentage:")
print(y.value_counts(normalize=True) * 100)
```

```python
# Categorical and numerical columns
categorical_features = ["state_name"]

numerical_features = [
    "rfs",
    "normal",
    "deviation",
    "month",
    "previous_error_mean",
    "previous_bust_rate"
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "state",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numeric",
            "passthrough",
            numerical_features
        )
    ]
)

# Random Forest
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                max_depth=12,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

# Chronological split: 80% earlier data, 20% later data
split = int(len(X) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("Training:", X_train.shape)
print("Testing :", X_test.shape)

# Train
model.fit(X_train, y_train)

print("\nMODEL TRAINING COMPLETE")
```

```python
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python
# Historical normal rainfall by state and month
monthly_normal = (
    rainfall
    .groupby(["state_name", rainfall["date"].dt.month])["normal"]
    .mean()
    .reset_index()
)

monthly_normal = monthly_normal.rename(
    columns={"date": "month"}
)

# Current NWP data
current = nwp_data.copy()

# Aggregate GFS grid values to state using our nearest grid mapping
current_state = state_df[
    ["state_name", "grid_latitude", "grid_longitude"]
].merge(
    current,
    left_on=["grid_latitude", "grid_longitude"],
    right_on=["latitude", "longitude"],
    how="left"
)

# Month of forecast
current_state["month"] = pd.to_datetime(
    current_state["valid_time"]
).dt.month

# Use GFS precipitation as forecast rainfall
current_state["rfs"] = current_state["tp"]

# Add historical normal rainfall
current_state = current_state.merge(
    monthly_normal,
    on=["state_name", "month"],
    how="left"
)

# Calculate rainfall deviation
current_state["deviation"] = np.where(
    current_state["normal"] > 0,
    ((current_state["rfs"] - current_state["normal"])
     / current_state["normal"]) * 100,
    0
)

# Add historical state error behavior
history_features = (
    hist.groupby("state_name")
    .agg(
        previous_error_mean=("forecast_error", "mean"),
        previous_bust_rate=("forecast_bust", "mean")
    )
    .reset_index()
)

current_state = current_state.merge(
    history_features,
    on="state_name",
    how="left"
)

print("Current forecast dataset:", current_state.shape)

current_state[
    [
        "state_name",
        "lead_day",
        "rfs",
        "temperature_C",
        "r2",
        "pressure_hPa",
        "wind_speed",
        "previous_error_mean",
        "previous_bust_rate"
    ]
].head()
```

```python
# Features expected by our trained model
X_current = current_state[
    [
        "state_name",
        "rfs",
        "normal",
        "deviation",
        "month",
        "previous_error_mean",
        "previous_bust_rate"
    ]
].copy()

# Handle missing values
X_current["normal"] = X_current["normal"].fillna(
    rainfall["normal"].median()
)

X_current["deviation"] = X_current["deviation"].fillna(0)

X_current["previous_error_mean"] = X_current[
    "previous_error_mean"
].fillna(hist["forecast_error"].median())

X_current["previous_bust_rate"] = X_current[
    "previous_bust_rate"
].fillna(hist["forecast_bust"].mean())

# Predict probability of forecast bust
current_state["bust_probability"] = (
    model.predict_proba(X_current)[:, 1]
)

print(
    current_state[
        ["state_name", "lead_day", "bust_probability"]
    ].head(20)
)
```

```python
# Normalize rainfall stress
rain_stress = (
    current_state["rfs"] /
    (current_state["normal"] + 1)
).clip(0, 3) / 3

# Historical error stress
error_stress = (
    current_state["previous_error_mean"] /
    hist["forecast_error"].quantile(0.95)
).clip(0, 1)

# Historical bust stress
bust_stress = current_state["previous_bust_rate"].clip(0, 1)

# Humidity stress
humidity_stress = (
    (current_state["r2"] - 60) / 40
).clip(0, 1)

# Wind stress
wind_stress = (
    current_state["wind_speed"] / 15
).clip(0, 1)

# Final Forecast Stress Score: 0–100
current_state["stress_score"] = (
    100 * (
        0.40 * current_state["bust_probability"] +
        0.20 * rain_stress +
        0.15 * error_stress +
        0.10 * bust_stress +
        0.10 * humidity_stress +
        0.05 * wind_stress
    )
).clip(0, 100)

# Confidence
current_state["confidence"] = pd.cut(
    current_state["stress_score"],
    bins=[-1, 30, 60, 100],
    labels=["HIGH", "MEDIUM", "LOW"]
)

print(
    current_state[
        [
            "state_name",
            "lead_day",
            "bust_probability",
            "stress_score",
            "confidence"
        ]
    ].head(20)
)
```

```python
# ============================================
# HISTORICAL FORECAST BUST TRAINING DATA
# ============================================

historical_ml = rainfall.dropna(
    subset=["actual", "rfs", "normal", "deviation"]
).copy()

historical_ml["date"] = pd.to_datetime(
    historical_ml["date"]
)

# Absolute forecast error
historical_ml["forecast_error"] = (
    historical_ml["actual"] -
    historical_ml["rfs"]
).abs()

# Forecast bust
historical_ml["bust"] = (
    historical_ml["forecast_error"] >= 20
).astype(int)

# Time features
historical_ml["month"] = (
    historical_ml["date"].dt.month
)

historical_ml["year"] = (
    historical_ml["date"].dt.year
)

# Historical rolling error
historical_ml = historical_ml.sort_values(
    ["state_name", "date"]
)

historical_ml["previous_error_mean"] = (
    historical_ml
    .groupby("state_name")["forecast_error"]
    .transform(
        lambda x:
        x.shift(1)
        .rolling(30, min_periods=5)
        .mean()
    )
)

historical_ml["previous_bust_rate"] = (
    historical_ml
    .groupby("state_name")["bust"]
    .transform(
        lambda x:
        x.shift(1)
        .rolling(30, min_periods=5)
        .mean()
    )
)

# Fill early-history missing values
historical_ml["previous_error_mean"] = (
    historical_ml["previous_error_mean"]
    .fillna(
        historical_ml["forecast_error"].median()
    )
)

historical_ml["previous_bust_rate"] = (
    historical_ml["previous_bust_rate"]
    .fillna(
        historical_ml["bust"].mean()
    )
)

print("Historical training rows:",
      len(historical_ml))

print("\nBust distribution:")
print(historical_ml["bust"].value_counts())

print("\nBust percentage:")
print(
    historical_ml["bust"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================
# TRAIN FORECAST BUST MODEL
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Features available BEFORE knowing actual rainfall
model_features = [
    "state_name",
    "rfs",
    "normal",
    "deviation",
    "month",
    "previous_error_mean",
    "previous_bust_rate"
]

X = historical_ml[model_features].copy()
y = historical_ml["bust"].copy()

# Chronological split
historical_ml = historical_ml.sort_values("date")

split = int(len(historical_ml) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))

print("\nTraining busts:", y_train.sum())
print("Testing busts :", y_test.sum())

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "state",
            OneHotEncoder(handle_unknown="ignore"),
            ["state_name"]
        )
    ],
    remainder="passthrough"
)

# Random Forest
rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                max_depth=12,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

# Train
rf_model.fit(X_train, y_train)

print("\nMODEL TRAINING COMPLETE!")
```

```python
# ============================================
# MODEL EVALUATION
# ============================================

y_pred = rf_model.predict(X_test)
y_probability = rf_model.predict_proba(X_test)[:, 1]

print("CLASSIFICATION REPORT")
print("=" * 50)

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)

print("CONFUSION MATRIX")
print("=" * 50)
print(confusion_matrix(y_test, y_pred))

print("\nROC-AUC:")
print(round(roc_auc_score(y_test, y_probability), 4))
```

```python
# ============================================
# CURRENT DAY 1-10 FORECAST FEATURES
# ============================================

current = nwp_data.copy()

# Attach each state to its nearest GFS grid point
current_state = state_df[
    ["state_name", "grid_latitude", "grid_longitude"]
].merge(
    current,
    left_on=["grid_latitude", "grid_longitude"],
    right_on=["latitude", "longitude"],
    how="left"
)

# Forecast month
current_state["month"] = pd.to_datetime(
    current_state["valid_time"]
).dt.month

# GFS precipitation forecast
current_state["rfs"] = current_state["tp"]

# Historical monthly normal rainfall
monthly_normal = (
    rainfall
    .assign(month=rainfall["date"].dt.month)
    .groupby(["state_name", "month"])["normal"]
    .mean()
    .reset_index()
)

current_state = current_state.merge(
    monthly_normal,
    on=["state_name", "month"],
    how="left"
)

# Fill missing normal rainfall
current_state["normal"] = current_state["normal"].fillna(
    rainfall["normal"].median()
)

# Forecast deviation from normal
current_state["deviation"] = np.where(
    current_state["normal"] > 0,
    (
        (current_state["rfs"] - current_state["normal"])
        / current_state["normal"]
    ) * 100,
    0
)

# Historical state behaviour
history_features = (
    historical_ml
    .groupby("state_name")
    .agg(
        previous_error_mean=("forecast_error", "mean"),
        previous_bust_rate=("bust", "mean")
    )
    .reset_index()
)

current_state = current_state.merge(
    history_features,
    on="state_name",
    how="left"
)

# Fill missing historical values
current_state["previous_error_mean"] = (
    current_state["previous_error_mean"]
    .fillna(historical_ml["forecast_error"].median())
)

current_state["previous_bust_rate"] = (
    current_state["previous_bust_rate"]
    .fillna(historical_ml["bust"].mean())
)

print("Current dataset shape:", current_state.shape)

print("\nLead days:")
print(
    current_state["lead_day"]
    .value_counts()
    .sort_index()
)

display(
    current_state[
        [
            "state_name",
            "lead_day",
            "rfs",
            "normal",
            "deviation",
            "temperature_C",
            "r2",
            "wind_speed",
            "previous_error_mean"
        ]
    ].head(10)
)
```

```python
# ============================================
# PREDICT FORECAST BUST PROBABILITY
# ============================================

model_features = [
    "state_name",
    "rfs",
    "normal",
    "deviation",
    "month",
    "previous_error_mean",
    "previous_bust_rate"
]

X_current = current_state[model_features].copy()

current_state["bust_probability"] = (
    rf_model.predict_proba(X_current)[:, 1]
)

print("Bust probability generated!")

display(
    current_state[
        [
            "state_name",
            "lead_day",
            "bust_probability"
        ]
    ].head(20)
)
```

```python
print(
    current_state["bust_probability"].describe()
)
```

```python
# ============================================
# UNIQUE FEATURE: FORECAST STRESS SCORE
# ============================================

# 1. ML predicted bust risk
ml_risk = current_state["bust_probability"]

# 2. Rainfall stress
rain_stress = (
    current_state["rfs"] /
    (current_state["normal"] + 1)
).clip(0, 3) / 3

# 3. Historical forecast-error stress
error_limit = historical_ml["forecast_error"].quantile(0.95)

error_stress = (
    current_state["previous_error_mean"] /
    error_limit
).clip(0, 1)

# 4. Historical bust frequency
historical_bust_stress = (
    current_state["previous_bust_rate"]
).clip(0, 1)

# 5. Humidity stress
humidity_stress = (
    (current_state["r2"] - 60) / 40
).clip(0, 1)

# 6. Wind stress
wind_stress = (
    current_state["wind_speed"] / 15
).clip(0, 1)

# ============================================
# FINAL SCORE
# ============================================

current_state["stress_score"] = (
    100 * (
        0.40 * ml_risk +
        0.20 * rain_stress +
        0.15 * error_stress +
        0.10 * historical_bust_stress +
        0.10 * humidity_stress +
        0.05 * wind_stress
    )
).clip(0, 100)

print("Forecast Stress Score created!")

display(
    current_state[
        [
            "state_name",
            "lead_day",
            "bust_probability",
            "stress_score"
        ]
    ].head(20)
)
```

```python
# ============================================
# FORECAST CONFIDENCE
# ============================================

current_state["confidence"] = pd.cut(
    current_state["stress_score"],
    bins=[-1, 30, 60, 100],
    labels=[
        "HIGH CONFIDENCE",
        "MEDIUM CONFIDENCE",
        "LOW CONFIDENCE"
    ]
)

print("Confidence distribution:")
print(
    current_state["confidence"]
    .value_counts()
)

display(
    current_state[
        [
            "state_name",
            "lead_day",
            "bust_probability",
            "stress_score",
            "confidence"
        ]
    ].head(20)
)
```

```python
# ============================================
# EXPLAINABLE FORECAST REASONS
# ============================================

def get_reasons(row):
    reasons = []

    if row["bust_probability"] >= 0.50:
        reasons.append("Very high ML bust risk")
    elif row["bust_probability"] >= 0.20:
        reasons.append("Elevated ML bust risk")

    if row["rfs"] > row["normal"] * 1.5:
        reasons.append("Rainfall forecast well above normal")

    if row["previous_error_mean"] > error_limit:
        reasons.append("High historical forecast error")

    if row["previous_bust_rate"] >= 0.20:
        reasons.append("Frequent historical forecast busts")

    if row["r2"] >= 90:
        reasons.append("Very high humidity")

    if row["wind_speed"] >= 10:
        reasons.append("Strong winds")

    if len(reasons) == 0:
        reasons.append("Weather conditions relatively stable")

    return " | ".join(reasons)


current_state["risk_reasons"] = current_state.apply(
    get_reasons,
    axis=1
)

display(
    current_state[
        [
            "state_name",
            "lead_day",
            "bust_probability",
            "stress_score",
            "confidence",
            "risk_reasons"
        ]
    ].head(20)
)
```
